```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

### Load Dataset

```python
iris = datasets.load_iris()
X = iris.data
y = iris.target
```

### Feature Standarization

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### Implement K-means Cluster with K=3 (based domain)

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)
```

```python
X_scaled
```

### Visualisasi Cluster with PCA (untuk mereduksi dimensi ke 2)

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

### label prediksi dari K-means

```python
y_kmeans = kmeans.predict(X_scaled)
```

```python
# Plot hasil clustering
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='viridis', label='Cluster')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroid')
plt.title("K-means Clustering on Iris Dataset (PCA Reduced)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend()
plt.show()
```
