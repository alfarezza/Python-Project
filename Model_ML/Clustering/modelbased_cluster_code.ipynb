## Model-Based Clustering

### 1. Gaussian Mixture Model (GMM)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
```

```python
# Membuat Data Dummy: 3 Klaster, masing-masing mengikuti distribusi Gaussian yang berbeda
np.random.seed(42)

# Klaster 1: Mean = (2, 2), Covariance = [[0.5, 0], [0, 0.5]], 200 titik
cluster_1 = np.random.multivariate_normal([2, 2], [[0.5, 0], [0, 0.5]], 200)

# Klaster 2: Mean = (7, 7), Covariance = [[1, 0], [0, 1]], 200 titik
cluster_2 = np.random.multivariate_normal([7, 7], [[1, 0], [0, 1]], 200)

# Klaster 3: Mean = (12, 2), Covariance = [[0.8, 0.2], [0.2, 0.8]], 200 titik
cluster_3 = np.random.multivariate_normal([12, 2], [[0.8, 0.2], [0.2, 0.8]], 200)
```

```python
# Gabungkan semua klaster menjadi satu dataset
data = np.vstack([cluster_1, cluster_2, cluster_3])
```

```python
data
```

```python
# Visualisasi Data Dummy
plt.figure(figsize=(8, 6))
plt.scatter(data[:, 0], data[:, 1], s=10, color='gray', alpha=0.6)
plt.title('Data Dummy (3 Klaster Gaussian)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
# Implementasi Gaussian Mixture Model (GMM)
gmm = GaussianMixture(n_components=3, covariance_type='full', random_state=42)
gmm.fit(data)
```

```python
# Prediksi Klaster
labels = gmm.predict(data)
```

```python
# Visualisasi Hasil Klastering
plt.figure(figsize=(8, 6))
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis', s=10)
plt.title('Hasil Klastering dengan GMM')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.colorbar(label='Klaster')
plt.show()
```

```python
# Menampilkan parameter dari GMM (Mean dan Covariance)
print("Means of the clusters:")
print(gmm.means_)

print("\nCovariance matrices of the clusters:")
print(gmm.covariances_)

print("\nWeights of the clusters:")
print(gmm.weights_)
```

#### Penjelasan Hasil:
- Visualisasi Data: Plot pertama menunjukkan distribusi data yang dibuat secara acak. Data terlihat seperti tiga kelompok yang terpisah.
- Visualisasi Hasil Klastering: Plot kedua menunjukkan bagaimana GMM mengelompokkan data tersebut. GMM berhasil mengidentifikasi dan memisahkan data ke dalam 3 klaster yang berbeda.

Parameter GMM:

- Means: Menunjukkan posisi rata-rata (centroid) dari masing-masing klaster.
- Covariances: Menunjukkan sebaran data dalam klaster (seberapa besar data tersebar di sekitar centroid).
- Weights: Menunjukkan seberapa besar proporsi masing-masing klaster dalam data keseluruhan.
