### 🔶 1. K-MEANS CLUSTERING

```python
import pandas as pd

data = pd.DataFrame({
    'Pendapatan': [15, 16, 18, 45, 50, 52, 85, 90, 88, 22, 68],
    'Pengeluaran': [16, 18, 14, 50, 55, 58, 90, 95, 92, 5, 77]
})
```

```python
data
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
# Data
X = data[['Pendapatan', 'Pengeluaran']]

# K-Means
kmeans = KMeans(n_clusters=3, random_state=0)
data['Cluster'] = kmeans.fit_predict(X)
```

```python
# Visualisasi
plt.scatter(data['Pendapatan'], data['Pengeluaran'], c=data['Cluster'], cmap='viridis')
plt.xlabel('Pendapatan')
plt.ylabel('Pengeluaran')
plt.title('K-Means Clustering')
plt.show()
```
