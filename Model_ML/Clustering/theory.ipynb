## 1. Partitional Clustering (Klastering Partisi)

### 📘 Apa itu Partitional Clustering?  
Partitional Clustering (klastering partisi) adalah salah satu pendekatan dalam unsupervised learning untuk mengelompokkan data ke dalam beberapa kelompok (klaster), dengan membagi data menjadi bagian-bagian yang tidak saling tumpang tindih (partisi).

> Tujuannya: Mencari cara terbaik untuk membagi data ke dalam sejumlah K kelompok, sehingga data dalam satu klaster mirip satu sama lain, dan berbeda dengan data di klaster lain.

### 🎯 Karakteristik Partitional Clustering:
- Setiap data hanya boleh masuk ke satu klaster (tidak tumpang tindih).
- Jumlah klaster (K) biasanya ditentukan di awal oleh pengguna.
- Proses ini berorientasi pada optimasi, seperti meminimalkan jarak dalam klaster.
- K-Means adalah algoritma paling terkenal dalam pendekatan ini.



### ⚙️ Cara Kerja Partitional Clustering (secara umum):
Walaupun cara kerjanya bisa berbeda tergantung algoritmanya, secara umum partitional clustering mengikuti langkah-langkah ini:

1. Pilih jumlah klaster (K).
Misalnya, kita ingin membagi data ke dalam 3 kelompok.
2. Inisialisasi pusat klaster (centroid) secara acak.
3. Tugaskan tiap data ke klaster terdekat. Biasanya berdasarkan jarak Euclidean.
4. Perbarui posisi pusat klaster berdasarkan rata-rata data yang masuk.
5. Ulangi proses assign dan update sampai konvergen (misalnya klaster tidak berubah lagi).

### ✨ Kelebihan Partitional Clustering:
Sederhana dan cepat untuk data besar.  
Efisien secara komputasi.  
Cocok untuk data dengan bentuk klaster yang jelas.

### ⚠️ Kekurangan:
Harus menentukan jumlah klaster (K) di awal.  
Tidak cocok untuk data dengan bentuk klaster kompleks (seperti bentuk melingkar).  
Sensitif terhadap inisialisasi awal (dalam algoritma seperti K-Means).

### 🤖 Algoritma

#### 🔶 1. K-Means Clustering
📘 Teori:
K-Means adalah algoritma klastering berbasis jarak yang digunakan untuk membagi data numerik ke dalam K klaster berdasarkan kedekatan dengan centroid (titik rata-rata dalam klaster). Tujuannya adalah untuk meminimalkan jarak total (variansi) antara data dan centroid-nya.
> Cocok untuk data numerik dan klaster berbentuk bulat/kompak.

#### ⚙️ Cara Kerja:
1. Tentukan jumlah klaster (K).
2. Inisialisasi centroid secara acak.
3. Tugaskan tiap data ke centroid terdekat (menggunakan jarak Euclidean).
4. Hitung ulang centroid sebagai rata-rata dari semua data dalam klaster.
5. Ulangi langkah 3 dan 4 sampai:
>Tidak ada perubahan klaster, atau  
>Mencapai iterasi maksimum.

✅ Cocok Untuk:  
Data numerik.  
Skala besar dengan bentuk klaster sederhana.



#### 🔷 2. K-Medoids Clustering (PAM – Partitioning Around Medoids)
📘 Teori:  
K-Medoids mirip dengan K-Means, tetapi lebih robust terhadap outlier. Alih-alih menggunakan centroid (rata-rata), K-Medoids menggunakan medoid (titik nyata dalam klaster yang paling representatif).
> Medoid adalah data aktual yang meminimalkan total jarak ke semua titik lain dalam klaster.

#### ⚙️ Cara Kerja:
1. Tentukan jumlah klaster (K).
2. Pilih K medoid awal secara acak dari data.
3. Tugaskan setiap data ke medoid terdekat (misalnya dengan jarak Manhattan).
4. Coba ganti medoid dengan titik lain, lalu hitung total cost (total jarak ke semua titik dalam klaster).
5. Jika cost berkurang, ganti medoid. Ulangi sampai tidak ada perbaikan.

✅ Cocok Untuk:
- Data dengan outlier.
- Dataset kecil hingga menengah.
- Jarak non-Euclidean (misal Manhattan, Hamming).

#### 🔶 3. K-Modes & K-Prototypes
📘 Teori:
K-Modes dan K-Prototypes adalah perluasan dari K-Means untuk menangani data kategorikal atau campuran (numerik + kategorikal).
- K-Modes: Untuk data kategorikal saja.
- K-Prototypes: Untuk data campuran (numerik + kategorikal).

#### K-Modes – Cara Kerja:
1. Tentukan jumlah klaster (K).
2. Inisialisasi mode (nilai paling sering) dari setiap fitur sebagai pusat klaster.
3. Tugaskan data ke klaster terdekat berdasarkan jumlah perbedaan kategori.
4. Hitung ulang mode baru dari fitur kategorikal untuk tiap klaster.
5. Ulangi proses sampai klaster stabil.
- Jarak digunakan: Matching dissimilarity (berapa banyak nilai berbeda antara dua data).

#### K-Prototypes – Cara Kerja:
Gabungkan:
- Jarak Euclidean untuk fitur numerik.
- Matching dissimilarity untuk fitur kategorikal.

Lalu:
1. Inisialisasi centroid yang berisi nilai rata-rata (numerik) dan mode (kategorikal).
2. Hitung jarak total gabungan untuk tiap data.
3. Tugaskan data ke klaster terdekat.
4. Update centroid: hitung mean dan mode masing-masing fitur.
5. Ulangi sampai konvergen.


✅ Cocok Untuk:
- K-Modes → data murni kategorikal (misal: warna, jenis kelamin, kota).
- K-Prototypes → data campuran (misal: umur + jenis pekerjaan).

| Algoritma    | Tipe Data   | Library            | Modul yang Digunakan |
| ------------ | ----------- | ------------------ | -------------------- |
| K-Means      | Numerik     | scikit-learn       | `KMeans`             |
| K-Medoids    | Numerik     | scikit-learn-extra | `KMedoids`           |
| K-Modes      | Kategorikal | kmodes             | `KModes`             |
| K-Prototypes | Campuran    | kmodes             | `KPrototypes`        |


## 2. Hierarchical Clustering (Klastering Hirarki)

Hierarchical Clustering adalah teknik **unsupervised learning** yang membentuk struktur hierarki (bertingkat) dari data dalam bentuk dendrogram (pohon). Algoritma ini tidak perlu menentukan jumlah klaster (K) di awal.
  
Terdapat dua pendekatan utama:

### 🔹 1. Agglomerative Clustering (Bottom-Up)

📘 Teori:  
Agglomerative Clustering adalah pendekatan bottom-up, artinya:
- Setiap data dimulai sebagai satu klaster sendiri-sendiri.
- Lalu, secara bertahap, klaster yang paling mirip digabung satu per satu.
- Proses berlanjut hingga hanya tersisa satu klaster besar (root dari pohon). 


Dari hasil ini, kita bisa memotong pohon pada level tertentu untuk mendapatkan jumlah klaster yang diinginkan.

### ⚙️ Cara Kerja (Langkah-langkah):
1. Setiap data dianggap sebagai satu klaster.
2. Hitung jarak antar semua klaster (pakai linkage — misalnya: **single, complete, average, ward**).
3. Gabungkan dua klaster yang paling dekat.
4. Update jarak antar klaster setelah penggabungan.
5. Ulangi langkah 2–4 sampai semua data tergabung dalam satu klaster.


📊 Contoh Linkage Criteria:
- Single linkage: Jarak antar dua titik terdekat.
- Complete linkage: Jarak antar dua titik terjauh.
- Average linkage: Rata-rata jarak antar semua pasangan titik.
- Ward linkage: Meminimalkan variansi dalam klaster.


✅ Cocok Untuk:
- Data numerik dan kategorikal (dengan modifikasi).
- Visualisasi struktur data (dendrogram).

### 🔸 2. Divisive Clustering (Top-Down)

📘 Teori:  
Divisive Clustering adalah pendekatan top-down, artinya:
- Mulai dari satu klaster besar berisi seluruh data.
- Lalu, secara bertahap, pecah klaster menjadi dua, berdasarkan perbedaan antar data.
- Proses ini terus dilakukan hingga tiap data menjadi klaster sendiri-sendiri atau hingga mencapai jumlah klaster tertentu.

### ⚙️ Cara Kerja (Langkah-langkah):
1. Mulai dari satu klaster besar (semua data).
2. Pilih klaster untuk dipecah (biasanya yang paling tidak homogen).
3. Pecah klaster menjadi dua, menggunakan metode seperti:
- K-Means
- Principal Component Analysis (PCA) + threshold
4. Ulangi proses pemecahan sampai kondisi berhenti (misalnya jumlah klaster yang diinginkan tercapai).



### 🔎 Perbedaan Utama:

| Aspek                 | Agglomerative    | Divisive                       |
| --------------------- | ---------------- | ------------------------------ |
| **Arah**              | Bottom-up        | Top-down                       |
| **Mulai dari**        | Klaster individu | Satu klaster besar             |
| **Proses**            | Penggabungan     | Pemecahan                      |
| **Kompleksitas umum** | Lebih efisien    | Lebih mahal (komputasi)        |
| **Populer?**          | Sangat populer   | Kurang umum (komputasi tinggi) |

### 📌 Kesimpulan Singkat:

| Algoritma     | Tipe      | Langkah Utama           | Kelebihan                              | Kekurangan                        |
| ------------- | --------- | ----------------------- | -------------------------------------- | --------------------------------- |
| Agglomerative | Bottom-up | Gabungkan klaster mirip | Visualisasi dendrogram, umum digunakan | Bisa lambat untuk data besar      |
| Divisive      | Top-down  | Pecah klaster besar     | Lebih akurat dalam beberapa kasus      | Komputasi mahal, jarang digunakan |

## 3. Density-Based Clustering

### Apa Itu Density-Based Clustering?
Density-based clustering adalah jenis klastering yang mengelompokkan data berdasarkan **kepadatan titik di ruang data**. Jadi, metode ini tidak hanya melihat jarak antar titik (seperti pada K-Means), tapi juga bagaimana titik data berada dalam kelompok yang padat.

### Cara Kerja Density-Based Clustering
Pada Density-Based Clustering, kita mengasumsikan bahwa klaster terbentuk di wilayah dengan kepadatan titik data yang tinggi, sementara area dengan sedikit titik dianggap sebagai noise atau outlier.
  
Berikut adalah langkah umum untuk algoritma density-based clustering:
1. Identifikasi Titik-Titik yang Padat: Titik yang berada dalam area dengan banyak tetangga dalam radius tertentu akan dianggap sebagai bagian dari klaster.
2. Ekspansi Klaster: Jika titik berada dalam wilayah padat (disebut juga core point), maka titik-titik yang berada di sekitar titik ini akan bergabung ke dalam klaster.
3. Noise atau Outlier: Titik yang tidak dapat dimasukkan ke dalam klaster mana pun dan tidak memiliki cukup tetangga di sekitarnya akan dianggap sebagai outlier (misalnya, titik yang jaraknya jauh dari klaster mana pun).

### Bagaimana Cara Kerja Density-Based Clustering? (Bahasa Awam)
Bayangkan kamu sedang berada di sebuah taman yang ramai. Orang-orang yang duduk bersama dalam kelompok besar adalah klaster. Kelompok yang lebih kecil atau orang yang berdiri sendiri adalah outlier.
- Core Points: Anggota dalam kelompok besar, yang dikelilingi oleh banyak orang dalam jarak yang sangat dekat. Mereka menjadi pusat dari kelompok tersebut.
- Border Points: Orang yang berada di tepi kelompok, masih cukup dekat dengan orang lain di kelompok, tetapi tidak dikelilingi oleh banyak orang.
- Noise: Orang yang berdiri sendiri, jauh dari kelompok besar, dan tidak dapat dikelompokkan.

### Jenis Algoritma Density-Based Clustering:
Ada beberapa algoritma yang menggunakan prinsip ini, yang paling terkenal adalah:

#### 1. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

#### 🧾 Teori DBSCAN:
DBSCAN mengelompokkan titik data berdasarkan kepadatan titik sekitar mereka. DBSCAN bekerja dengan dua parameter utama:
- ε (epsilon): Radius dari sebuah titik yang digunakan untuk mendeteksi tetangga (lingkaran di sekitar titik).
- MinPts: Jumlah titik minimal dalam radius ε untuk membentuk sebuah klaster.

#### ⚙️ Cara Kerja DBSCAN:
1. Titik Core: Titik yang memiliki setidaknya MinPts tetangga dalam radius ε dianggap sebagai titik core.
2. Ekspansi Klaster: Titik core yang memiliki banyak tetangga akan menambah titik-titik tetangganya ke dalam klaster yang sama.
3. Titik Border: Titik yang berada di sekitar titik core, tetapi tidak memiliki cukup tetangga untuk menjadi titik core, akan dimasukkan ke dalam klaster.
4. Noise/Outlier: Titik yang tidak memiliki cukup tetangga dan tidak berada dalam radius ε dari titik core akan dianggap sebagai noise.

#### 2. OPTICS (Ordering Points To Identify the Clustering Structure)

#### 🧾 Teori OPTICS:
OPTICS adalah algoritma clustering yang mirip dengan DBSCAN, tetapi dengan perbedaan utama dalam **cara klaster dibentuk dan diatur**. OPTICS tidak memerlukan jumlah klaster yang pasti di awal. Sebagai gantinya, ia memberikan informasi lebih lengkap tentang **struktur kedalaman klaster berdasarkan kepadatan**.

#### ⚙️ Cara Kerja OPTICS:
1. Titik Core: Sama seperti DBSCAN, OPTICS juga mencari titik core berdasarkan ε dan MinPts.
2. Ekspansi Klaster: OPTICS lebih fleksibel dalam memperhitungkan kedalaman dan struktur klaster, memungkinkan klaster dengan kepadatan yang berbeda untuk tumpang tindih.
3. Hasil: OPTICS memberikan urutan kedalaman untuk setiap titik yang menggambarkan bagaimana kedekatan mereka dengan titik lain dalam struktur klaster.


OPTICS tidak perlu menentukan jumlah klaster di awal, karena ia memberikan informasi yang lebih komprehensif tentang kepadatan dan hubungan antar klaster.

#### Ringkasan Perbedaan DBSCAN vs OPTICS:

| Fitur               | DBSCAN                                                       | OPTICS                                                  |
| ------------------- | ------------------------------------------------------------ | ------------------------------------------------------- |
| **Parameter Utama** | `eps` (radius), `min_samples` (titik minimal)                | `min_samples` (titik minimal)                           |
| **Jumlah Klaster**  | Ditetapkan sebelumnya (kita memilih `eps` dan `min_samples`) | Tidak perlu jumlah klaster yang ditentukan di awal      |
| **Kelebihan**       | Cepat dan efisien untuk data dengan noise                    | Menangani kepadatan yang berbeda dengan lebih fleksibel |
| **Kekurangan**      | Sensitif terhadap parameter `eps`                            | Lebih kompleks dalam implementasi dan interpretasi      |

#### ✅ Kesimpulan:

- DBSCAN sangat efektif untuk mengidentifikasi **klaster yang terpisah** dengan jelas dan **mengabaikan noise**.

- OPTICS lebih fleksibel dan memberi **struktur klaster** yang lebih mendalam tanpa memerlukan jumlah klaster yang pasti di awal.

Keduanya cocok untuk data yang memiliki **noise atau klaster** dengan bentuk yang tidak beraturan.
