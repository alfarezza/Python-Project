### 🤖 Machine Learning

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

#### Create Func Evaluation Model

```python
def evaluate_model(model, X_test, y_test, name='Model'):
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    print(f"📊 {name}")
    print(f"MAE  : {mae:,.3f}")
    print(f"RMSE : {rmse:,.3f}")
    print(f"R²   : {r2:.4f}")
    print("-" * 30)
    
    return y_pred
```

#### ✅ Train & Evaluate Model 

##### Linear Reg

```python
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_preds = evaluate_model(lr_model, X_test, y_test, name='Linear Regression')
```

##### Random Forest Reg

```python
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
rf_preds = evaluate_model(rf_model, X_test, y_test, name='Random Forest')
```

##### XGBoost Reg

```python
xgb_model = XGBRegressor(n_estimators=100, random_state=42)
xgb_model.fit(X_train, y_train)
xgb_preds = evaluate_model(xgb_model, X_test, y_test, name='XGBoost')
```

#### 📊 Visualization Prediction vs Actual

```python
def plot_preds(y_test, y_pred, model_name):
    plt.figure(figsize=(6,6))
    sns.scatterplot(x=y_test, y=y_pred, alpha=0.6)
    plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--', color='red')
    plt.xlabel('Actual Price')
    plt.ylabel('Predicted Price')
    plt.title(f'{model_name} - Actual vs Predicted')
    plt.tight_layout()
    plt.show()
```

```python
# Contoh visualisasi
plot_preds(y_test, lr_preds, "Linear Regression")
plot_preds(y_test, rf_preds, "Random Forest")
plot_preds(y_test, xgb_preds, "XGBoost")
```

#### 📈 Metric Interpretation
- MAE (Mean Absolute Error): Average prediction error.
- RMSE (Root Mean Squared Error): Similar to MAE, but larger penalty for extreme errors.
- R² (R-squared): How much of the target (Price) variation is explained by the model (max = 1.0).

#### 🧠 Analysis
**Linear Regression:** <br>
🔻 MAE and RMSE are very high<br>
🔻 R² is only 0.786 → meaning the model only explains ~79% of price variation.<br>
❌ Not suitable for non-linear and complex data such as property data.<br>

**Random Forest:** <br>
✅ Best overall performance<br>
✅ Very low MAE and RMSE<br>
✅ R² close to 1 → almost all price variation is very well explained.<br>

**XGBoost:** <br>
⚡ Also very good, only slightly below Random Forest.<br>
🔺 Faster for big data and supports advanced fine-tuning.<br>

#### ✅ Conclusion
- 🏆 Random Forest is the best model at the moment, with the most accurate house price predictions.
- XGBoost is also very good and suitable if you want further tuning or integration into production systems.
- Linear Regression is too simple for complex property datasets and is not recommended.

### 🎯 Hyperparameter Tuning dengan GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
```

#### ✅Tuning Random Forest

```python
rf_params = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['log2', 'sqrt']
}
```

```python
rf = RandomForestRegressor(random_state=42)
rf_grid = GridSearchCV(estimator=rf,
                       param_grid=rf_params,
                       cv=3,
                       n_jobs=-1,
                       scoring='neg_root_mean_squared_error',
                       verbose=1)

rf_grid.fit(X_train, y_train)

print("📊 Best Parameters for Random Forest:")
print(rf_grid.best_params_)
print(f"Best RMSE: {-rf_grid.best_score_:.3f}")
```

#### ✅ Tuning XGBoost

```python
xgb_params = {
    'n_estimators': [100, 200],
    'max_depth': [3, 6, 10],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.7, 1.0],
    'colsample_bytree': [0.7, 1.0]
}
```

```python
# 🔍 GridSearch untuk XGBoost

xgb = XGBRegressor(random_state=42, objective='reg:squarederror')
xgb_grid = GridSearchCV(estimator=xgb,
                        param_grid=xgb_params,
                        cv=3,
                        n_jobs=-1,
                        scoring='neg_root_mean_squared_error',
                        verbose=1)

xgb_grid.fit(X_train, y_train)

print("📊 Best Parameters for XGBoost:")
print(xgb_grid.best_params_)
print(f"Best RMSE: {-xgb_grid.best_score_:.3f}")
```

```python
best_rf = rf_grid.best_estimator_
best_xgb = xgb_grid.best_estimator_
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_model(model, X_test, y_test, name='Model'):
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    print(f"📊 {name} (After Tuning)")
    print(f"MAE  : {mae:,.0f}")
    print(f"RMSE : {rmse:,.0f}")
    print(f"R²   : {r2:.4f}")
    print("-" * 30)
    return y_pred
```

```python
best_rf = rf_grid.best_estimator_
rf_tuned_preds = evaluate_model(best_rf, X_test, y_test, name='Random Forest (Tuned)')
```

```python
best_xgb = xgb_grid.best_estimator_
xgb_tuned_preds = evaluate_model(best_xgb, X_test, y_test, name='XGBoost (Tuned)')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_preds(y_test, y_pred, model_name):
    plt.figure(figsize=(6, 6))
    sns.scatterplot(x=y_test, y=y_pred, alpha=0.5)
    plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], '--', color='red')
    plt.xlabel('Actual Price')
    plt.ylabel('Predicted Price')
    plt.title(f'{model_name} - Actual vs Predicted')
    plt.tight_layout()
    plt.show()

# Plot hasil akhir
plot_preds(y_test, rf_tuned_preds, "Random Forest (Tuned)")
plot_preds(y_test, xgb_tuned_preds, "XGBoost (Tuned)")
```
