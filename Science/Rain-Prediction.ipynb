## 🔹Outlier Detection with IQR (Interquartile Range)

```python
# Fungsi deteksi outlier
def detect_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return outliers
```

```python
# Contoh: cek outlier di 'MaxTemp'
outlier_maxtemp = detect_outliers_iqr(df_model, 'MaxTemp')
print(f"Jumlah outlier MaxTemp: {len(outlier_maxtemp)}")
```

```python
for col in numeric_cols:
    outlier_count = len(detect_outliers_iqr(df_model, col))
    print(f"{col}: {outlier_count} outlier")
```
