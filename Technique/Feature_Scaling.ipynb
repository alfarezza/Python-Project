## 🔄 Apa Itu Feature Scaling?

Feature Scaling adalah proses **mengubah skala (range) dari nilai-nilai fitur (kolom) numerik dalam dataset**, agar model machine learning bisa bekerja lebih efisien dan akurat.<br>

📌 Beberapa model sangat sensitif terhadap skala fitur, seperti:
- K-Nearest Neighbors (KNN)
- Support Vector Machine (SVM)
- Logistic Regression
- PCA
- Neural Network


📈 Kenapa Scaling Itu Penting? (Contoh Nyata) <br>
Bayangkan kamu sedang menganalisis data pasien:<br>
- Usia (range: 18–90)
- Tekanan Darah (range: 80–180)
- Pendapatan (range: 2 juta – 100 juta)

Jika kamu tidak melakukan scaling, maka fitur seperti Pendapatan akan mendominasi perhitungan jarak di model seperti KNN atau SVM, padahal mungkin Usia lebih penting untuk menentukan penyakit.

## ⚙️ Tiga Teknik Scaling Populer:

### ✅ 1. StandardScaler (Standarisasi Z-score)

Mengubah data agar punya mean = 0 dan standar deviasi = 1  
📌 Cocok untuk: model linear seperti logistic regression, SVM, PCA.

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

# Data: [Usia, Pendapatan]
data = np.array([[25, 20000000],
                 [40, 30000000],
                 [60, 80000000]])
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

print("Standard Scaled Data:\n", scaled_data)
```

### ✅ 2. MinMaxScaler (Normalisasi ke Range [0, 1])

Mengubah nilai ke dalam rentang 0 sampai 1 (atau range lain jika ditentukan)  
📌 Cocok untuk: data yang sudah diketahui batas atas/bawahnya, neural network.

```python
from sklearn.preprocessing import MinMaxScaler # import library
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)

print("MinMax Scaled Data:\n", scaled_data)
```

### ✅ 3. RobustScaler (Tahan terhadap Outlier)

Gunakan median dan IQR (interquartile range) alih-alih mean/std → lebih stabil jika ada outlier ekstrem.  
📌 Cocok untuk: dataset yang mengandung nilai ekstrem / pencilan

```python
from sklearn.preprocessing import RobustScaler
```

```python
data_1 = np.array([[25, 20000000],
                 [40, 30000000],
                 [60, 80000000],   # outlier
                 [35, 21000000]])
```

```python
scaler = RobustScaler()
scaler_data = scaler.fit_transform(data_1)
```

```python
print("Robust Scaled Data:\n", scaler_data)
```

## 🧠 Kapan Menggunakan Teknik yang Mana?

| Teknik             | Cocok Untuk                                | Hindari Jika                 |
| ------------------ | ------------------------------------------ | ---------------------------- |
| **StandardScaler** | Data sebarannya normal, model linear       | Banyak outlier               |
| **MinMaxScaler**   | Neural nets, data tidak mengandung outlier | Ada outlier ekstrem          |
| **RobustScaler**   | Ada outlier, ingin skala stabil            | Tidak butuh outlier handling |
