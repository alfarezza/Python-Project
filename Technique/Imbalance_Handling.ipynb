```python
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
from collections import Counter
```

#### Create Inbalance Dataset

```python
X, y = make_classification(n_samples=1000, n_features=2, n_informative=2, 
                           n_redundant=0, n_clusters_per_class=1,
                           weights=[0.9, 0.1], random_state=42)
```

```python
X
```

```python
# Lihat distribusi awal
print(f"Distribusi sebelum balancing: {Counter(y)}")
```

#### Visualize Distribution

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', alpha=0.6)
plt.title("Distribusi Data Sebelum Balancing")
plt.show()
```

#### 🔄 Oversampling with SMOTE

```python
from imblearn.over_sampling import SMOTE
```

```python
# Terapkan SMOTE
smote = SMOTE(random_state=42)
X_smote, y_smote = smote.fit_resample(X, y)

print(f"Distribusi setelah SMOTE: {Counter(y_smote)}")
```

```python
# Visualisasi hasil SMOTE
plt.scatter(X_smote[:, 0], X_smote[:, 1], c=y_smote, cmap='bwr', alpha=0.6)
plt.title("Distribusi Data Setelah SMOTE (Oversampling)")
plt.show()
```

#### 📉 Undersampling with RandomUnderSampler

```python
from imblearn.under_sampling import RandomUnderSampler
```

```python
# Terapkan undersampling
rus = RandomUnderSampler(random_state=42)
X_rus, y_rus = rus.fit_resample(X, y)
```

```python
print(f"Distribusi setelah undersampling: {Counter(y_rus)}")
```

```python
# Visualisasi hasil undersampling
plt.scatter(X_rus[:, 0], X_rus[:, 1], c=y_rus, cmap='bwr', alpha=0.6)
plt.title("Distribusi Data Setelah Random Undersampling")
plt.show()
```
