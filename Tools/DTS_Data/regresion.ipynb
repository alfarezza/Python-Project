```python
import numpy as np
from sklearn.linear_model import LinearRegression
import  matplotlib.pyplot as plt
```

```python
# Input data
hours_studied = np.array([1, 2, 3, 4, 5, 6, 7]).reshape(-1, 1)  # Independent variable (X)
final_scores = np.array([50, 55, 60, 65, 70, 75, 78])  # Dependent variable (Y)
```

```python
# Create and fit model
model = LinearRegression()
model.fit(hours_studied, final_scores)
```

```python
# Predict for 8 hours of study
predicted_score = model.predict([[8]])
```

```python
# Model parameters
slope = model.coef_[0]
intercept = model.intercept_
```

```python
# Plotting the results
plt.scatter(hours_studied, final_scores, color='blue')
plt.plot(hours_studied, model.predict(hours_studied), color='red')
plt.title('Simple Linear Regression: Hours Studied vs Final Score')
plt.xlabel('Hours Studied')
plt.ylabel('Final Score')
plt.show()
```

```python
print(f"Slope (m): {slope}")
print(f"Intercept (b): {intercept}")
print(f"Predicted score for 8 hours of study: {predicted_score[0]}")
```
